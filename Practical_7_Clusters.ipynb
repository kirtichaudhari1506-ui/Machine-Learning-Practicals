```python
#import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.rcParams['figure.figsize'] = (6,5)  #Default fig size
plt.rcParams['figure.dpi'] = 150  #Default fig resolution
sns.set_style('darkgrid')  #Default diagram theme
```

```python
df = pd.read_csv('Mall_Customers.csv')
df.head()
```

```python
df.info()
```

```python
# Drop the unnecessary columns
df.drop(['CustomerID', 'Gender', 'Age'], axis=1, inplace=True)
df.head()
```

```python
#Data visualization
sns.scatterplot(x=df['Annual Income (k$)'],
                y=df['Spending Score (1-100)'])
plt.show()
```

```python
#Standardization
from sklearn.preprocessing import StandardScaler

sc = StandardScaler()

x = sc.fit_transform(df)
x[:10]
```

```python
from sklearn.cluster import KMeans

km = KMeans()  # default 8 Cluster
yp = km.fit_predict(x)

yp
```

```python
len(yp)
```

```python
#cluster centre
km.cluster_centers_
```

```python
#Visualize the clusters- Scatter plot With Centroid of Each Cluster
plt.scatter(x[:,0], x[:,1], c=yp, cmap='viridis')
plt.scatter(km.cluster_centers_[:,0], km.cluster_centers_[:,1],
            c='black', s=200)

#C - Centroid Colour  S - Size of Centroid 100 to 200 range
plt.show()
```

```python
km.inertia_  # sum of squared distances of each sample to its nearest cluster centre
```

```python
#Elbow curve is the plot between K values and WCSS
#WCSS (Within-Cluster Sum of Squares)

wcss = []

for i in range(1,20):
    km = KMeans(n_clusters=i)
    km.fit_predict(x)
    wcss.append(km.inertia_)

wcss
```

```python
plt.plot(range(1,20), wcss)
plt.xlabel('K-Values')
plt.ylabel('WCSS')
plt.title('Elbow Graph')
plt.show()
```

```python
from sklearn.metrics import silhouette_score

silhouette_score(x, yp)
```

```python
# Step 5: Silhouette Score Analysis
silhouette_scores = []

for k in range(2, 11):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = kmeans.fit_predict(x)
    score = silhouette_score(x, labels)
    silhouette_scores.append(score)

    print(f"k = {k}, Silhouette Score = {score:.3f}")
```

```python
# Visualize Silhouette Scores
plt.figure(figsize=(7, 4))
plt.plot(range(2, 11), silhouette_scores, marker='o')
plt.title('Silhouette Score Method')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Silhouette Score')
plt.xticks(range(2, 11))
plt.grid(True)
plt.show()
```
